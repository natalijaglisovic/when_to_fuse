```python
def plot_modality_drop(datasets, dataset_names):
    """Plot 1x3 subplots for modality drop across datasets."""
    fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

    methods = [
        ('add', 'Addition', 'o', '#2ecc71'),
        ('concat', 'Concatenation', 's', '#3498db'),
        ('attention', 'Attention', '^', '#e74c3c'),
    ]

    legend_handles = []
    for ax, data, name in zip(axes, datasets, dataset_names):
        summary = data['summary']
        percentages = [10, 20, 30, 40, 50, 60, 70, 80, 90]

        for method, label, marker, color in methods:
            means = [summary[method][f'{p}%']['mean'] for p in percentages]
            line, = ax.plot(percentages, means, label=label, marker=marker,
                           color=color, linewidth=2, markersize=6)
            if name == dataset_names[0]:
                legend_handles.append(line)

        ax.set_xlabel('Drop Rate (%)', fontsize=12)
        ax.set_title(name, fontsize=14)
        ax.grid(True, alpha=0.3)

    axes[0].set_ylabel('Recall@20', fontsize=12)

    fig.legend(handles=legend_handles, labels=[m[1] for m in methods],
              loc='upper center', ncol=3, fontsize=12, frameon=False,
              bbox_to_anchor=(0.5, 1.05))

    plt.tight_layout()
    #plt.savefig('modality_drop_subplots.png', bbox_inches='tight')
    plt.show()

plot_modality_drop(
    [amazon_baby, amazon_games, amazon_science],
    ['Amazon Baby', 'Amazon Games', 'Amazon Science']
)
```
